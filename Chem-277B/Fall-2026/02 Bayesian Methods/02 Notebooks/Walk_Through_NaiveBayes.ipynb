**2) Plotting the Data**<br>
<br>

Let us now plot the data by using UMAP as a dimension reduction method. **You will learn more about UMAP soon.** For now, the only important information about UMAP is that it is a projection method that maps high dimensional data to a 2D or 3D space so that we can visualize it. The mapping causes some distortions, but we still get a good idea about the data distribution.<br> 
Before we run any analysis, including UMAP, we need to scale the data. Scaling and normalizing the data is important because the features are measured in arbitrary units. Depending on the unit system, the numerical values of the features change, but it is still the same feature. Therefore, we usually normalize the features all together to a certain range (usually zero to one) or scale it to a standard normal if the features are sufficiently normal distributed. 

````python
scaler   = StandardScaler() 
XTrainS  = scaler.fit_transform(XTrain)
````

Next, we run the UMAP transformation from 5D to 3D and then generate a scatter plot. 

````python
fit    = UMAP(n_components = 3)
XTrans = fit.fit_transform(XTrainS)
````

Plotting in 3D:

````python
colorsIdx = {'Non-Toxic': 'black', 'Toxic': 'red'}
cols      = YTrain.map(colorsIdx)

scatter = go.Scatter3d(x = XTrans[:,0], y = XTrans[:,1], z = XTrans[:,2], mode = 'markers', marker = dict(size = 5, color = cols))
fig     = go.Figure(data = [scatter])
fig.update_layout(width = 800, height = 800, margin = dict(r = 10, b = 10, l = 100, t = 10))
fig.show()
````

Let us also compare the scaled data to the raw data:

**3) Naive Bayes**<br>
<br>

We are now running Naive Bayes on the scaled training data set.

````python
gnb = GaussianNB()            # initializing Gaussian Naive Bayes
Fit = gnb.fit(XTrainS, YTrain)# fitting the model
````

In the next step, we want to predict the classes from the test dataset using our model and thereby evaluate the quality of the model. Note, that we run ```scaler.transform``` and **not** ```scaler.fit_transform``` on the test set! Running ```scaler.fit_transform``` on the test set would scale to the test sets mean and variance (and not to the training sets mean and variance) and would prevent the model from generalizing to new data. 

````python
#extracting X and Y from the test data set
XTest = Test.drop('label', axis = 1).values
YTest = Test['label']
````

````python
#scaling the test set
XTestS = scaler.transform(XTest)
````

````python
Ypred = Fit.predict(XTestS)        #predicting classes
Probs = Fit.predict_proba(XTestS)  #calculating probabilities for classes
````

````python
print(Probs[:10,:])
````

````python
print(Ypred[:10])
````

We see that the first column (index = 0) of *Probs* refers to "Non-Toxic" and the second column (index = 1) refers to "Toxic"
